```python
string='abhiraj'
count=0
for i in string:
    if i in 'aeiou':
        count+=1
print(count)
```

```python
def vowels(string):
    result=''
    for i in string:
        if i not in 'aeiou':
            result+=str(i)
    return result
print(vowels('abhiraj1'))
```
