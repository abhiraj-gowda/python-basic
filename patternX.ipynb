```python
c=7
for i in range (1,c+1):
    for j in range(1,c+1):
        if i==1 or i==c or j==1 or j==c or i==j or i+j==c+1:
            print('*',end=' ')
        else:
            # pass
            print('',end="  ")
    print()
```
